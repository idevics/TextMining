```python
import pandas as pd
```

```python
data = pd.read_csv('data/per.csv', encoding='utf8') # encoding='utf8'-> چون فارسی
data.head()
```

```python
with open('data/stopwords.txt', encoding='utf8') as stopwords_file:
    stopwords = stopwords_file.readlines()


stopwords = [line.replace('\n', '') for line in stopwords]
stopwords
```

```python
import nltk
# nltk.download('stopwords')
```

```python
nltk_stopwords = nltk.corpus.stopwords.words('english')
stopwords.extend(nltk_stopwords)
len(stopwords)
```

```python
import hazm
stemmer = hazm.Stemmer()
```

```python
stemmer.stem('کتاب ها')
# ریشه ی کلمه رو برمیگردونه
```

```python
from hazm import word_tokenize

word_tokenize('سلام چطور هستی ؟')
```

```python
sample_data = data.sample(n=100, random_state=42)
dataset = pd.DataFrame(columns=('title_body', 'category'))
for index, row in sample_data.iterrows():
    title_body = f" {row['Title']} {row['Body']} "
    title_body_tokenized = word_tokenize(title_body)
    title_body_tokenized_filtered = [w for w in title_body_tokenized if not w in stopwords]
    title_body_tokenized_filtered_stemmed = [stemmer.stem(w) for w in title_body_tokenized_filtered]
    dataset.loc[index] = {
        'title_body': ' '.join(title_body_tokenized_filtered_stemmed),
        'category': row['Category2'].replace('\n', '').replace('\u200c', '')
    }
```

```python
dataset.head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
vectorizer = TfidfVectorizer()
vectorizer.fit(dataset['title_body'])
```

```python
import pickle
with open('lion_v.jsdh', 'wb') as f:
    pickle.dump(vectorizer, f)
```

```python
X = vectorizer.transform(dataset['title_body'])
```

```python
X
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
y = le.fit_transform(dataset['category'])
```

```python
le.classes_
```

```python
len(y)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y)
```

```python
from sklearn import svm
```

```python
svmc = svm.SVC()
svmc.fit(X_train, y_train)
```

```python
svmc.score(X_test, y_test)
# هرچقدر تعداد کلاس ها بره بالا دقت بالا گرفتن سخت تر میشه
```

```python
from sklearn.metrics import classification_report, confusion_matrix
y_pred = svmc.predict(X_test)
```

```python
print(classification_report(y_test, y_pred))
```

```python
print(confusion_matrix(y_test, y_pred))
```
